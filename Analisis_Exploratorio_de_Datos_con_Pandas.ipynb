##Importando librerias

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

##Creando DataFrame

```python
df = pd.read_csv('/content/world_population.csv')
```

```python
df
```

```python
pd.set_option('display.float_format', lambda x: '%.2f' % x)
```

##Verificando tipos de dato

```python
df.info()
```

##Primeras estadisticas

```python
df.describe()
```

##Buscando Nulos

```python
df.isnull().sum()
```

##Filas unicas

```python
df.nunique()
```

```python
df.sort_values(by='2022 Population', ascending = False).head(10)
```

##Buscando correlaciones

```python
df.corr(numeric_only = True)
```

```python
sns.heatmap(df.corr(numeric_only = True), annot = True)

plt.rcParams['figure.figsize'] = (20,7)
plt.show()
```

##Agrupando por Continente

```python
df.groupby('Continent').mean(numeric_only = True).sort_values(by='2022 Population', ascending = False)
```

##Visualizando historia por continente

```python
df2 = df.groupby('Continent')[['1970 Population',
       '1980 Population', '1990 Population', '2000 Population',
       '2010 Population', '2015 Population', '2020 Population',
       '2022 Population']].mean(numeric_only = True).sort_values(by='2022 Population', ascending = False)
```

```python
df.columns
```

```python
df3 = df2.transpose()
df3
```

```python
df3.plot()
```

##Boxplot para ver outliers

```python
df.boxplot(figsize = (18,10))
```
